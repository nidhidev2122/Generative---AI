```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer
```

```python
# Load the GPT-2 tokenizer and model
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")
```

```python
# Common sentence used for comparing different generation settings
prompt = "The future of technology depends on the choices we make today."

tokens = tokenizer(prompt, return_tensors="pt")
```

## basic generation

```python
result = model.generate(
    **tokens,
    max_new_tokens=50
)

text = tokenizer.decode(result[0], skip_special_tokens=True)
print(text)
```

##Experiment 1 — Low Temperature

```python
result = model.generate(
    **tokens,
    max_new_tokens=80,
    do_sample=True,
    temperature=0.3
)

print(tokenizer.decode(result[0], skip_special_tokens=True))
```

##Experiment 2 — High Temperature

###Temperature = 1.5

###High temperature flattens the probability distribution. The model picks more unusual tokens, making the output more creative but sometimes random or incoherent.

```python
result = model.generate(
    **tokens,
    max_new_tokens=80,
    do_sample=True,
    temperature=1.5
)

print(tokenizer.decode(result[0], skip_special_tokens=True))
```

##Experiment 3 — Small Top-k

Top-k = 5

```python
result = model.generate(
    **tokens,
    max_new_tokens=80,
    do_sample=True,
    temperature=1.0,
    top_k=5
)

print(tokenizer.decode(result[0], skip_special_tokens=True))
```

##Experiment 4 — Low Top-p

Top-p = 0.5

```python
result = model.generate(
    **tokens,
    max_new_tokens=80,
    do_sample=True,
    temperature=1.0,
    top_k=0,
    top_p=0.5
)

print(tokenizer.decode(result[0], skip_special_tokens=True))
```

##Experiment 5 — Repetition Penalty

Repetition penalty = 2.0

```python
result = model.generate(
    **tokens,
    max_new_tokens=80,
    do_sample=True,
    temperature=1.0,
    repetition_penalty=2.0
)

print(tokenizer.decode(result[0], skip_special_tokens=True))
```

#Summary Table — Five Experiments
| Experiment | Setting | Value | Result |
|------------|---------|-------|--------|
| 1 | Temperature | 0.3 | More predictable and repetitive |
| 2 | Temperature | 1.5 | More random and creative |
| 3 | Top-k | 5 | Restricted and focused vocabulary |
| 4 | Top-p | 0.5 | Focused generation with less variety |
| 5 | Repetition penalty | 2.0 | Reduces repeated words and phrases |


#Custom Prompt

```python
story_prompt = "She opened the old wooden door and found a letter that changed everything."

story_tokens = tokenizer(story_prompt, return_tensors="pt")
```

##Generation 1 — Balanced

```python
result = model.generate(
    **story_tokens,
    max_new_tokens=80,
    do_sample=True,
    temperature=0.7,
    top_k=50,
    top_p=0.95,
    repetition_penalty=1.2
)

print(tokenizer.decode(result[0], skip_special_tokens=True))
```

##Generation 2 — Creative

```python
result = model.generate(
    **story_tokens,
    max_new_tokens=80,
    do_sample=True,
    temperature=1.2,
    top_k=0,
    top_p=0.95,
    repetition_penalty=1.0
)

print(tokenizer.decode(result[0], skip_special_tokens=True))
```

##Generation 3 — Controlled

```python
result = model.generate(
    **story_tokens,
    max_new_tokens=80,
    do_sample=True,
    temperature=0.5,
    top_k=40,
    top_p=1.0,
    repetition_penalty=1.5,
    pad_token_id=tokenizer.eos_token_id
)

print(tokenizer.decode(result[0], skip_special_tokens=True))
```

#Final Analysis

Temperature: The temperature setting produced one of the clearest differences. With a low value such as 0.3, the generated text was more predictable and tended to repeat ideas. At 1.5, the output had greater variety, although some sentences became less meaningful.

Top-k: Using top_k=5 restricted the possible choices considerably. The output was more controlled because the model selected from a small group of likely tokens.

Top-p: With top_p=0.5, the model focused mainly on high-probability choices. This resulted in text with less variation compared with settings using a larger probability range.

Repetition penalty: A value of 2.0 helped reduce repeated words and phrases. However, using a very high penalty can sometimes make the generated sentences sound less natural.

Overall: The experiments show that generation parameters have a significant effect on GPT-2 output. Lower or more restrictive settings generally make the text more controlled, while higher randomness allows more variation but can reduce coherence.
